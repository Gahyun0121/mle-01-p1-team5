1. 파일 불러오기

```python
import pandas as pd

cd  = pd.read_csv("../data/processed/country_dashboard.csv")
al  = pd.read_csv("../data/processed/alarm_final.csv")
inc = pd.read_csv("../data/processed/incident_final.csv")
sn  = pd.read_csv("../data/processed/safety_notice_final.csv")
de  = pd.read_csv("../data/destinations_clean.csv")

iso3 = "PHL"
```

2. 상단 3칸

```python
cd["iso3"] == iso3
cd[cd["iso3"] == iso3]
row = cd[cd["iso3"] == iso3].iloc[0]
print(row)

print(row["국가명"], row["경보단계"], row["안전공지_총건수"])

iso2 = row["ISO 코드"]
print(iso2)
```

3. 대표 경보

```python
a = al[al["iso3"] == iso3].sort_values("경보단계", ascending=False)

top = a.iloc[0]
print(f"{top['경보단계']}단계 · {top['경보내용']}")
```

4. 지역별 경보 전체

```python
# iterrows: 표를 한 줄씩 돌려줌. (인덱스, 행) 두 개가 나옴
for _, r in a.iterrows():
    print(r["경보단계"], r["경보내용"])
```

5. 사건사고 태그 (등장횟수 순)

```python
KEYWORDS = ["소매치기","강도","절도","납치","성범죄","마약","사기",
            "테러","시위","교통사고","분실","폭행","자연재해","감염병","살인"]

text = str(inc.loc[inc["ISO코드"] == iso2, "사건사고내용_clean"].iloc[0])

# text.count("강도") → 그 단어가 몇 번 나오는지 세는 문자열 메서드
counts = {k: text.count(k) for k in KEYWORDS if text.count(k) > 0}

# key=counts.get 값(횟수)을 기준
tags = sorted(counts, key=counts.get, reverse=True)[:5]
print(tags)
```

6. 최신 공지 5개

```python
news = sn[sn["ISO코드"] == iso2] \
         .sort_values("안전공지_작성일", ascending=False) \
         .head(5)

print(news[["안전공지_작성일", "공지제목"]])
```

7. 추천 정보

```python
d = de[de["iso3"] == iso3]

if d.empty:
    print("추천 정보 없음")
else:
    # .explode()     → 리스트 한 칸을 여러 행으로 펼침
    # set()          → 중복 제거
    themes = sorted(set(d["themes"].str.split("|").explode()))
    months = sorted(set(d["months"].str.split("|").explode()))
    print(themes, months)
```
